```python
# 1. importar as bibliotecas
import pandas as pd
import seaborn as srn
import matplotlib.pyplot as plt # Adicionando import para pyplot
import statistics as sts

# 2. Importar o arquivo
dataset = pd.read_csv("Churn.csv", sep=";")

# 3. Renomear as colunas
dataset.columns= ["Id", "Score", "Estado", "Genero", "Idade", "Patrimonio", "Saldo", "Produtos", "TemCartCredito", "Ativo", "Salario", "Saiu"]
dataset.head()


#Salário
j.fig.suptitle("Salario")
```

```python
dataset['Genero'], dataset['Genero'].isnull().sum()
```

```python
dataset.isnull().sum() # Problema doNulo de Salario corrigidos
```

```python
#Correção dos dados M, F e Fem
dataset.loc[dataset['Genero'] == 'M', 'Genero'] = "Masculino"
dataset.loc[dataset['Genero'].isin(['F','Fem']), 'Genero'] = "Feminino"

#Visualizar
agrupado = dataset.groupby(["Genero"]).size() # Agrupa por Gênero
agrupado
```

```python
dataset.loc[(dataset['Idade'] < 0) | (dataset['Idade'] > 120)] # veriica as idade fora de 0 e 120 anos
mediana_idade = sts.median(dataset['Idade'])
```

```python
mediana_idade # menos sujeios a aoutliers - fora do padrão
dataset.loc[(dataset['Idade'] < 0) | (dataset['Idade'] > 120),'Idade'] = mediana_idade #- Substituição pela mediana
dataset.loc[(dataset['Idade'] < 0) | (dataset['Idade'] > 120)]# verificaçõ pelo filto
```

```python
#Amostragem Sistemática
import pandas as pd
import numpy as np
from math import ceil

#Leitura do banco de dados
populacao = 150
amostra = 15
k = ceil(populacao/amostra)
print(k)
```

```python
#Def. do valor ramdômico para inicializar a amostra de 1 ate k
r = np.random.randint(low=1, high=k+1, size=1)
print(r)
```

```python
# Criar um for para a criação dos outros valores
acumulador = r[0]
sorteados = []
for i in range(amostra):
  sorteados.append(acumulador)
  acumulador += k
print(sorteados)
```

ARIMA e SARIMA

1. Parte Não-Sazonal: $(p, d, q)

    1. $p$ (Auto-Regressivo - AR): Quantidade de passos para atrás
    2. $d$ (Diferenciação - I): o número de subtrações do valor atual
    3. $q$ (Média Móvel - MA) : O ajuste da quantidade de erros das previsões anteriores.

2. Parte Sazonal: $(P, D, Q, s)$

    1. $s$ (Sazonalidade): Comprimento do ciclo;
    2. $P$ (AR Sazonal): A quantidade de ciclos completos para atrás;
    3. $D$ (Diferenciação Sazonal): Substração do valor atual pelo do mês e ano anterior;
    4. $Q$ (MA Sazonal): Ajuste do modelo com base no mesmo período do mês passado

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

# s=12 indica que o padrão se repete a cada 12 meses
# Os valores (1, 1, 1) sazonais abaixo são um bom ponto de partida

p = 4
d = 1
q = 3
modelo = SARIMAX(train, order=(p, d, q), seasonal_order=(1, 1, 1, 12))
modelo_treinado = modelo.fit()
```

```python
# Gere a previsão novamente
future_forecast = modelo_treinado.forecast(steps=len(test))
plt.plot(test)
plt.plot(future_forecast, color='red')
```

```python
pd.concat([data, future_forecast], axis=1).plot(linewidth=3)
```

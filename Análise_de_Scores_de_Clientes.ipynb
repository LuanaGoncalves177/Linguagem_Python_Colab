```python
#Análise de clientes
#Nota de crédito Score de Clientes
  #Bom (Good), okay(standart) ou ruim (poor)

# Passo a passo do projeto
  # 0. Entender o desafio da empresa: finir o objetivo final do projeo
  # 1. Importar a base e dados para tratamento

#pip install pandas scikit-learn - terminal

import pandas as pd

tabela =pd.read_csv("clientes.csv")
display(tabela)
display(tabela.info())
```

```python
  # 2. Tratar a base de dados para as IAs: Preparar o padrão - prever a coluna Score de crédito

    # Tratar na tabelas o valores que são testos (transfomar em números)
      #profissao
from sklearn.preprocessing import LabelEncoder

codificador_profissao = LabelEncoder()
tabela["profissao"] = codificador_profissao.fit_transform(tabela['profissao'])

      # mix_credito
codificador_mix_credito = LabelEncoder()
tabela["mix_credito"] = codificador_mix_credito.fit_transform(tabela['mix_credito'])

      #comportamento_pagamento
codificador_comportamento_pagamento = LabelEncoder()
tabela["comportamento_pagamento"] = codificador_comportamento_pagamento.fit_transform(tabela['comportamento_pagamento'])

display(tabela)
# inversão números para textos - Não inverte os valores
# codificador_profissao.inverse_transform( y=tabela["profissao"])
# display(tabela)
      # precisão da coluna "Score_credito"

```

Here's how you can use `inverse_transform` correctly to see the original text values for 'profissao', 'mix_credito', and 'comportamento_pagamento':

```python
# Como criar um proceso de IA
  # prever a coluna de "score_credito"

#y: O que se quer prever - "score_credito"
y = tabela["score_credito"]

#x: usar para fazer a previsão
x = tabela.drop(columns=["score_credito", "id_cliente"])

# Dvisão ds dados para texte e previsão
from sklearn.model_selection import train_test_split
x_treino, x_textest, y_treino, y_teste = train_test_split(x,y, test_size=0.3)
#70/30
```

```python
  # 3. Criar os modelos de previsão -> ruim, okay ou bom
#3.1. Importar a IA (Modelo)
# Árvore e descisão -> RandonFlorest e # KNN -> K Nearest Neighbors ( visinhos próximos)
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

#3.2. Criar a IA (Modelo)
modelo_arvore_decisao =RandomForestClassifier()
modelo_knn = KNeighborsClassifier()

#3.1. Treinar a IA (Modelo)
modelo_arvore_decisao.fit(x_treino, y_treino)
modelo_knn.fit(x_treino, y_treino)
```

```python
  # 4. Avaliar e escolher o melhor modelo

  # 4.1 Previsão
previsao_arvore = modelo_arvore_decisao.predict(x_textest)
previsao_knn = modelo_knn.predict(x_textest)

  # 4. 2 Comparação entre a pevião e o y_teste
from sklearn.metrics import accuracy_score
```

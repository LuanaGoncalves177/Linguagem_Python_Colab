Desafio 1: Função do cáulo da amplitude
*diferencia em tre o maior e o menor valor de um vetor

```python

def amplitude(lst):
  amp = print(max(lst) - min(lst))
  return(amp)


lst = [2, 6, 8, 11, 23, 35]
amplitude(lst)
```

Desafio 2: Imprimir uma string na vertical

```python
def vertical(palavra):
  return print(*palavra, sep = "\n")

palavra = input("Digite uma palavra: ")
vertical(palavra)
```

```python
peso = input("Digite o peso da bagagem:")
peso = float(peso)

if peso <= 10:
  print("Valor da bagagem: R$ 50,00")
elif peso >= 11 and peso <= 20:
  print("Valor da bagagem: R$ 80,00")
else:
  print("O trasporte não é aceito")
  print("\n favor, dividir a carga e realizar os envios separados")
```

Desafio 4: Calculadora(autoral)

```python
def calculadora(n1, n2, op):
  if op == "+":
    print(n1 + n2)
  elif op == "-":
    print(n1 - n2)
  elif op == "*":
    print(n1 * n2)
  elif op == "/":
    if n2 == 0:
      print("Não é possível dividir por zero")
    else:
      print(n1 / n2)
  else:
    print("Operação inválida")

n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
op = input("Digite a operação desejada (+, -, *, /): ")
calculadora(n1, n2, op)
```

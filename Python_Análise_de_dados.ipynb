```python
#Análise de dados com Python
#1 Passo a passo do projeto
  # 1. Abri a base de dados (importar a base de dados);
    #pandas: pip install pandas openpyxl nbformat ipykernel plotly

import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")

#print(tabela) imprimi a tabele desformatada

display(tabela)
  # 2. Visualizar a base de dados;
    # 2.1 entender as informações disponíveis;
    # 2.2 possívesis erros na base de dados;
    # 2.3 informações que não ajudam  que atrapalham
    # lista_nomes = ["lira", "talia"] criação de lista em python

print("Tabela limpa.")
tabela = tabela.drop(columns="CustomerID")
display(tabela)
```

```python
# 3.corrigir os prolemas da base de dados; (tratamentos de dados)
display(tabela.info())
  # 3.1 valores em formatos errados

  # 3.2 Informações vazias (Excluir informações)
print("Tabela limpa!")
tabela = tabela.dropna()
display(tabela.info())
```

```python
# 4. Análise inicial(entender o motivo do cancelamentos);
# 4.1 Quantidade
#contar quantos cara são 0 e 1 na coluna Candelou

print("Valores brutos cancelados")
display(tabela["cancelou"].value_counts())
#display(tabela["duracao_contrato"].value_counts()) - quantidade os planos

# 4.2 Porcentagem
print("Valores em percentuais")
display(tabela["cancelou"].value_counts(normalize=True))
#display(tabela["cancelou"].value_counts(normalize=True).map("{:.1%}".format)) formatar casas decimais
```

```python
# 5. Análise detalhada para enter a causa do cancelamento dos clientes.
import plotly.express as px
print(tabela.columns) # printa as colunas da tabela

# 5.1 Cria o gráfico
for coluna in tabela.columns:
  grafico = px.histogram(tabela, x=coluna, color="cancelou",text_auto=True)
  # 5.2 Mostra o gráfico
  grafico.show()
```
